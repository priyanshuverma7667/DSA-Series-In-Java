# Stemming & Lemitization

```python
import nltk
from nltk.stem import PorterStemmer
```

```python
stemmer = PorterStemmer()

words = ["Running","runner","ran","easily","fair"]

s = [stemmer.stem(word) for word in words]

print(s)
```
